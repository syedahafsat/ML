```python
# AI-Powered Interview Question Generator
# ---------------------------------------

import random
from transformers import pipeline

# --------------------------
# Step 1: Question Bank (sample dataset)
# --------------------------
technical_questions = {
    "Data Science": [
        "Explain the difference between supervised and unsupervised learning.",
        "What is overfitting and how can you prevent it?",
        "How would you handle missing data in a dataset?",
        "Explain precision, recall, and F1 score."
    ],
    "Software Engineering": [
        "Explain the concept of object-oriented programming.",
        "What is the difference between a stack and a queue?",
        "How do you handle version control in a team?",
        "What are design patterns and why are they useful?"
    ]
}

behavioral_questions = [
    "Tell me about a time you faced a challenge in a team project. How did you handle it?",
    "Describe a situation where you had to quickly learn something new.",
    "Give an example of when you showed leadership skills.",
    "Tell me about a time you failed and what you learned from it."
]

# --------------------------
# Step 2: Rule-based Generator
# --------------------------
def generate_interview_set(role, num_tech=2, num_behav=2):
    tech_qs = random.sample(technical_questions.get(role, []), num_tech)
    behav_qs = random.sample(behavioral_questions, num_behav)
    return tech_qs + behav_qs

# Example usage (rule-based)
print("=== Rule-based Interview Question Set ===")
role = "Data Science"
for q in generate_interview_set(role):
    print("-", q)

# --------------------------
# Step 3: AI-powered Generator (Hugging Face GPT-2/3 style models)
# --------------------------
print("\n=== AI-powered Interview Question Generation ===")

# Load pre-trained text generation pipeline
generator = pipeline("text-generation", model="gpt2")

def ai_generate_questions(role, profile, num_q=3):
    prompt = f"Generate {num_q} technical and behavioral interview questions for an intern applying for a {role} role. Intern profile: {profile}."
    results = generator(prompt, max_length=100, num_return_sequences=1)
    return results[0]['generated_text']

# Example usage (AI-powered)
role = "Software Engineering"
profile = "Python, SQL, teamwork experience"
print(ai_generate_questions(role, profile, 3))
```
