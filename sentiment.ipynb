```python
# Sentiment Analysis of Internship Feedback
# ----------------------------------------
# Using Logistic Regression (baseline) and Transformers (advanced)

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Sample dataset (you can replace with actual intern feedback CSV/Excel file)
data = {
    "feedback": [
        "The internship was a great learning experience.",
        "I felt undervalued and ignored during the internship.",
        "The team was supportive and helpful.",
        "The workload was too much with little guidance.",
        "I really enjoyed working on real projects.",
        "The management was not responsive to concerns."
    ],
    "sentiment": ["positive", "negative", "positive", "negative", "positive", "negative"]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Split into train/test
X_train, X_test, y_train, y_test = train_test_split(df["feedback"], df["sentiment"], test_size=0.3, random_state=42)

# TF-IDF Vectorization
vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Logistic Regression Model
log_reg = LogisticRegression()
log_reg.fit(X_train_vec, y_train)

# Predictions
y_pred = log_reg.predict(X_test_vec)
print("=== Logistic Regression Results ===")
print(classification_report(y_test, y_pred))


# ---------------------------------------------------------
# Transformer Model (BERT) for Sentiment Classification
# ---------------------------------------------------------
from transformers import pipeline

# Load pre-trained sentiment analysis pipeline (HuggingFace)
sentiment_model = pipeline("sentiment-analysis")

print("\n=== Transformer Model Results (BERT) ===")
for feedback in df["feedback"]:
    result = sentiment_model(feedback)[0]
    print(f"Feedback: {feedback}\nPredicted Sentiment: {result['label']} (score: {result['score']:.2f})\n")
```
